```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# read in output vmr file
df = pd.read_csv('output/grid_day/1200K_EL3_day/vmr_interp.dat', comment='#', sep="\\s+")
df2 = pd.read_csv('output/grid_day/1400K_EL3_day/vmr_interp.dat', comment='#', sep="\\s+")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
ax1.plot( df["T[K]"], df["P[bar]"], color="orange")
ax1.set_yscale("log")
ax1.set_xlabel("Temperature [K]")
ax1.set_ylabel("Pressure [bar]")
ax1.set_ylim(1e-3, 1e-7)

for col in df.columns:
    if not "[" in col:
        if df[df["P[bar]"] < 1e-3][col].max() > 1e-5:
            line = ax2.plot(df[col], df["P[bar]"], label=col)
            color = line[0].get_color()
            max_idx = df[df["P[bar]"] < 1e-3][col].idxmax()
            ax2.annotate(col, (df[col].iloc[max_idx], df["P[bar]"].iloc[max_idx]), color=color)

ax2.set_xscale("log")
ax2.set_xlabel("VMR")
ax2.set_xlim(1e-6, 1)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
ax1.plot( df2["T[K]"], df2["P[bar]"], color="orange")
ax1.set_yscale("log")
ax1.set_xlabel("Temperature [K]")
ax1.set_ylabel("Pressure [bar]")
ax1.set_ylim(1e-1, 1e-5)

for col in df2.columns:
    if not "[" in col:
        if df2[df2["P[bar]"] < 1e-2][col].max() > 1e-5:
            line = ax2.plot(df2[col], df2["P[bar]"], label=col)
            color = line[0].get_color()
            max_idx = df2[df2["P[bar]"] < 1e-2][col].idxmax()
            ax2.annotate(col, (df2[col].iloc[max_idx], df2["P[bar]"].iloc[max_idx]), color=color, fontsize=14)

ax2.set_xscale("log")
ax2.set_xlabel("VMR")
ax2.set_xlim(1e-6, 1)
```
